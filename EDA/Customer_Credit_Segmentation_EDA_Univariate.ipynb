# **Import Requirements**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from scipy import stats
```

```python
data = pd.read_csv("../data/processed/credit_card_clean.csv")
```

# **EDA Univariate**

## **Balance/Saldo**

```python
data_bal = data['balance']
data_bal.describe()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
  data_bal,
  bins = 30,
  color = 'darkblue',
  kde=True
)

plt.title("Distribusi Balance")
plt.xlabel("Balance")
plt.ylabel("Customer")
plt.show()
```

```python
skew = stats.skew(data_bal)
print("Skew : ", skew)
```

Right Skewed di Balance ➡ Mayoritas customer memiliki balance yang rendah

```python
print("Jumlah customer dengan balance 0 = ", len(data_bal[data_bal == 0]))
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(
    x=data_bal,
    color='skyblue'
)


plt.title('Boxplot Balance')
plt.xlabel('Balance')
plt.show()
```

Boxplot menunjukkan ➡ Terdapat potential outlier yang mengartikan terdapat juga dengan balance yang jauh lebih tinggi, namun bukan berarti outlier merupakan anomali atau kesalahan data.
