**Imputasi**

Fill Missing Values -> **minimum_payments**

```python
data.loc[
    (data["balance"] == 0) & (data["minimum_payments"].isna()),
    "minimum_payments"
] = 0

data[data["balance"] == 0].sample(10)
```

```python
missing = pd.DataFrame({
    "missing_count": data.isna().sum(),
    "missing_pct": data.isna().mean() * 100
})
display(
    missing.sort_values("missing_pct", ascending=False)
)
```

```python
med_min_payment = data["minimum_payments"].median()

data.loc[
     (data["balance"] > 0) & (data["minimum_payments"].isna()),
    "minimum_payments"
] = med_min_payment

data[data["minimum_payments"] == med_min_payment]
```

```python
missing = pd.DataFrame({
    "missing_count": data.isna().sum(),
    "missing_pct": data.isna().mean() * 100
})
display(
    missing.sort_values("missing_pct", ascending=False)
)
```

Fill Missing Values -> **credit_limit**

```python
data[data["credit_limit"].isna()][["cust_id", "credit_limit"]]
```

```python
med_credit_limit = data["credit_limit"].median()

data.loc[
    (data["credit_limit"].isna()), "credit_limit"
] = med_credit_limit

data[data["cust_id"] == "C15349"][["cust_id","credit_limit"]]
```
